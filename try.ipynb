```python
with open('input.txt', 'r', encoding='utf8') as f:
    text = f.read()
```

```python
print('Lenth of datset in characters', len(text))
```

```python
print('first 100 characters', text[:1000])
```

```python
#determine all characters
chars = sorted(list(set(text)))
vocab_size = len(chars)
print(''.join(chars))
print(vocab_size)
```

```python
#create mapping of characters to numbers (tokenizer)
stoi = { ch:i for i, ch in enumerate(chars)}
itos = { i:ch for i, ch in enumerate(chars)}
encode = lambda s: [stoi[c] for c in s] #encodes characters to numbers
decode = lambda s: ''.join([itos[c] for c in s]) #decodes numbers to characters
# there are many other different tokenizers but this is just for simple purpose

print(encode('hallo'))
print(decode(encode('hallo')))
```

```python
#encode entire dataset
import torch
data = torch.tensor(encode(text), dtype=torch.long)

print(data.shape, data.dtype)
print(data[:1000])
```

```python
#seperate into train and val set
n = int(0.9*len(data))
train_data = data[:n]
val_data = data[n:]
```

```python
#defining max length for modell to learn at once
batch_size = 8
train_data[:batch_size+1] #von 8 folgen das nächste dann predicten
```
